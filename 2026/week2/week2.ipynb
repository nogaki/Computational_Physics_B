## for文
本文4ページ

字下げは1段4個の空白で揃えます。この例では、ループ終了後も `i` の値は `4` です。

```python
n = 5
for i in range(n):
    print(i)
```

## for文の例：0から4を表示する
本文5ページ

`range(5)` から取り出す整数は0〜4です。終点の5は含みません。

```python
for i in range(5):
    print(i)
```

## rangeの開始値と刻み幅
本文6ページ

`range(start, stop, step)` は開始値・終点・刻み幅を指定します。終点は含まず、刻み幅を省略すると `1` です。負の刻み幅なら減少します。`0` は指定できません。

次の3つのセルで、開始値を1にする場合、2ずつ進む場合、逆順の場合を比べましょう。


```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(0, 10, 2):
    print(i)
```

```python
for i in range(5, 0, -1):
    print(i)
```

## for文の例：総和を求める
本文7ページ

1〜10を順に足します。$\sum_{i=1}^{n}i=\frac{n(n+1)}{2}$ を $n=10$ で確かめましょう。最後の `print(total)` はループの後で実行します。

```python
total = 0
for i in range(10):
    total = total + (i + 1)
print(total)
```

## 字下げで出力が変わる
本文8ページ

変えるのは `print(total)` の字下げだけです。字下げした行はループの中で繰り返し、字下げを戻した行はループが終わってから実行します。出力を予想してから実行しましょう。


### ループの内側
足すたびに途中経過を表示します。出力は1、3、6の3回です。


```python
total = 0
for i in range(1, 4):
    total = total + i
    print(total)
```

### ループの外側
全部足してから、最後の結果6を1回だけ表示します。


```python
total = 0
for i in range(1, 4):
    total = total + i
print(total)
```

## for文の例：二重ループ
本文9ページ

$\sum_{i=1}^{3}\sum_{j=1}^{5}ij$ を計算します。コードの `i + 1` は1〜3、`j + 1` は1〜5です。

```python
total = 0
for i in range(3):
    for j in range(5):
        total = total + (i+1)*(j+1)
print(total)
```

## if文
本文10ページ

条件を上から調べ、最初に真になった分岐だけを実行します。`n` を正、0、負に変えて試しましょう。

```python
n = 0
if n > 0:
    print("positive")
elif n == 0:
    print("zero")
else:
    print("negative")
```

## bool型の演算
本文12ページ

`and` は両方が真なら真、`or` は少なくとも一方が真なら真です。以下は、上から `(True, True)`、`(True, False)`、`(False, True)`、`(False, False)` の順です。

```python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

## if文の例：両方が偶数
本文13ページ

`%` は割り算のあまりです。`n % 2 == 0` で偶数かを判定し、`and` で両方が真かを調べます。

```python
n = 5
m = 10
if n % 2 == 0 and m % 2 == 0:
    print("Both are even.")
else:
    print("At least one is odd.")
```

## if文の例：少なくとも一方が偶数
本文14ページ

`or` は少なくとも一方が真なら真です。両方が偶数でも真になることを、`n = 4` に変えて確かめましょう。

```python
n = 5
m = 10
if n % 2 == 0 or m % 2 == 0:
    print("At least one is even.")
else:
    print("Both are odd.")
```

## for文とif文の組み合わせ
本文15ページ

0〜9のうち、偶数だけを表示します。`print(i)` は `if` の内側へさらに1段字下げします。

```python
for i in range(10):
    if i % 2 == 0:
        print(i)
```

## while文
本文16ページ

毎回、処理の前に条件を調べます。条件が変わらないと無限ループになることがあります。停止しないときはセルの停止ボタンを押します。

```python
n = 1
while n < 5:
    print(n)
    n = n + 1
```

## while文の例：2の累乗
本文17ページ

2を繰り返し2倍し、初めて100以上になる値を求めます。128になった時点で `n < 100` は偽になります。

```python
n = 2
while n < 100:
    n = n * 2
print(n)
```
