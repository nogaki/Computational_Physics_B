### 1. 名前と基本的な違い（配布PDF p.19）

| 種類 | 主な用途・特徴 | 次元と要素 |

### 2. 作り方と型の指定（配布PDF p.20）

最初のコードセルで標準ライブラリとNumPyを読み込みます。`array.array`の型コード`'d'`はCの`double`、`'i'`はCの符号付き`int`を表します。`'i'`が何bitかは実行環境に依存するため、常に32bitとは限りません。

NumPyの`np.array`は入力から`dtype`を推定できます。必要なら`dtype=float`のように明示します。`typecode`は要素型を指定するコード（この例では'd'や'i'）、`dtype`はNumPyのデータ型オブジェクトなので、同じ指定方法ではありません。出力では`.typecode`と`.dtype`、`.shape`を見比べます。

`np.zeros`と`np.ones`の既定の`dtype`は浮動小数点型です。配列を作るすべての関数が同じ既定値を持つわけではありません。

関連：[9. 整数型への小数代入](#appendix-a9)

```python
import array as ar
import numpy as np

standard_array = ar.array("d", [1.0, 2.0])
numpy_array = np.array([1.0, 2.0], dtype=float)
print(standard_array.typecode)
print(numpy_array.dtype, numpy_array.shape)
```

### 3. 同じ演算子でも意味が変わる（配布PDF p.21）

同じ型コードの標準`array`同士で`+`を使うと、値ごとの加算ではなく並びの連結になります。`array * 整数`は並び全体を繰り返します。一方、`ndarray`の`+`、`*`、`**`は対応する要素ごとの加算・乗算・累乗です。したがって同じ`[1, 2]`に対して、標準`array`の`* 2`は`[1, 2, 1, 2]`、NumPyの`* 2`は`[2, 4]`になります。

NumPyの行列積には`@`を使います。標準`array`の`**`はこのシーケンス演算として定義されていないため、ここでは実行せず説明にとどめます。

関連：[補足Cの要素ごとの演算・行列積](#numpy-elementwise)

```python
standard_array = ar.array("d", [1.0, 2.0])
numpy_array = np.array([1.0, 2.0])
print((standard_array + standard_array).tolist())
print(numpy_array + numpy_array)
print((standard_array * 2).tolist())
print(numpy_array * 2)
print(numpy_array ** 2)
```

### 4. 標準arrayの追加・削除（配布PDF p.22）

| 操作 | 役割 | 元の配列 |
|---|---|---|
| `append(x)` | 末尾へ1要素追加 | その場で変更 |
| `extend(items)` | iterableの要素を末尾へ追加 | その場で変更 |
| `insert(i, x)` | 添字`i`の位置へ追加 | その場で変更 |
| `pop([i])` | 添字の要素を取り除いて返す。省略時は末尾 | その場で変更 |
| `remove(x)` | 値`x`と最初に一致する要素を1つ削除 | その場で変更 |
| `del a[i]` | 添字`i`の要素を削除 | その場で変更 |

1つ目の例では`pop()`が末尾の`4.0`を取り除いて返し、その後`remove(2)`が最初の`2.0`だけを削除します。2つ目では`insert(1, 9)`の後に`del standard_values[2]`を行うため、その時点の添字2にある要素が消えます。空配列や範囲外の添字に対する`pop`は`IndexError`、存在しない値への`remove`は`ValueError`です。

関連：[5. NumPyの追加・削除](#appendix-a5)

```python
standard_values = ar.array("d", [1, 2, 2, 3])
standard_values.append(4)
print(standard_values.pop())
standard_values.remove(2)
print(standard_values.tolist())
```

```python
standard_values = ar.array("d", [1, 2])
standard_values.extend([3, 4])
standard_values.insert(1, 9)
del standard_values[2]
print(standard_values.tolist())
```

### 5. NumPyの追加・削除（配布PDF pp.23–24）

`ndarray`には`array.array`のような`.append()`、`.pop()`、`.remove()`メソッドはありません。NumPyには`np.append`、`np.insert`、`np.delete`という関数があり、処理結果の新しい配列を返します。この呼び出しでは元の配列を変更しないので、更新後の値が必要なら`numpy_values = np.append(numpy_values, 4)`のように返り値を受け取ります。要素代入`numpy_values[0] = ...`は可能です。

`np.delete(numpy_values, 1)`の`1`は削除する添字です。値が`2`の要素をすべて除きたい場合は`numpy_values[numpy_values != 2]`のように条件で選びます。これは最初に一致した1件だけを削除する`array.remove(2)`とは動作が異なります。

関連：[6. 多次元の追加とaxis](#appendix-a6) · [補足Cの条件・where](#numpy-conditions)

```python
numpy_values = np.array([1.0, 2.0, 2.0, 3.0])
grown_values = np.append(numpy_values, 4)
print(numpy_values)
print(grown_values)
print(np.delete(numpy_values, 1))
print(numpy_values[numpy_values != 2])
```

```python
numpy_values = np.array([1.0, 2.0])
inserted_values = np.insert(numpy_values, 1, 9)
print(numpy_values)
print(inserted_values)
```

### 6. 多次元の追加とaxis（配布PDF p.25）

`np.append`、`np.insert`、`np.delete`は`axis`を省略すると、入力を1次元に平坦化して処理します。この例では、2行2列の`matrix_positions`と1行2列の`row_position`を`np.append(matrix_positions, row_position)`で結合すると、計6要素の1次元配列となり、shapeは`(6,)`です。行を追加してshapeを`(3, 2)`に保つには、例のように`np.concatenate((matrix_positions, row_position), axis=0)`を使います。`np.concatenate`の既定`axis`は0です。

2次元配列では`axis=0`が行方向、`axis=1`が列方向の結合です。結合する軸以外の大きさは一致させる必要があります。出力のshapeを確認すると、配列の次元が保たれたかを見分けられます。

関連：[8. shape・reshape・転置・行列積](#appendix-a8)

```python
matrix_positions = np.array([[1, 0], [0, 2]])
row_position = np.array([[-1, 0]])
joined_positions = np.concatenate((matrix_positions, row_position), axis=0)
flattened_positions = np.append(matrix_positions, row_position)
print(joined_positions)
print(joined_positions.shape, flattened_positions.shape)
```

### 7. 切り出した配列はコピーか共有か（配布PDF pp.26–27）

標準`array`のスライスは独立したコピーです。NumPyの基本スライス（例：`a[1:3]`）は元のメモリを共有するviewなので、そのスライスへの代入が元配列にも反映されます。切り出した数値を独立させたい場合は`.copy()`を使います。

整数配列や条件配列を使った高度なインデックスの結果はコピーです。例では`selected_values = numpy_values[[1, 2]]`で取り出した配列を書き換えても、元の`numpy_values`は変わりません。なお、どちらの配列でも`b = a`は同じ配列オブジェクトへの別名を作ります。一方、`numpy_values[[1, 2]] = ...`のように添字へ直接代入すれば、元の`numpy_values`を書き換えます。

関連：[10. 相互変換とコピー](#appendix-a10)

```python
standard_values = ar.array("d", [1, 2, 3, 4])
standard_slice = standard_values[1:3]
standard_slice[0] = 99
print(standard_values.tolist())
print(standard_slice.tolist())
```

```python
numpy_values = np.array([1.0, 2.0, 3.0, 4.0])
numpy_slice = numpy_values[1:3]
numpy_slice[0] = 99
print(numpy_values)
numpy_copy = numpy_values[1:3].copy()
numpy_copy[0] = -1
print(numpy_values)
```

```python
numpy_values = np.array([1.0, 2.0, 3.0, 4.0])
selected_values = numpy_values[[1, 2]]
selected_values[0] = 99
print(numpy_values)
```

### 8. shape・reshape・転置・行列積（配布PDF p.28）

`ndarray`の`shape`は各軸の要素数を表します。標準`array`は1次元ですが、`ndarray`なら行・列・時刻などを複数の軸として扱えます。`reshape`は要素数を保ったまま形を変えた配列を返します。メモリの並び方などによりviewになる場合とcopyになる場合があるため、常にどちらかとは限りません。

2次元配列の`.T`は行と列を入れ替えます。`@`は行列積を計算します。1次元配列には行・列の向きがないため、`.T`を使ってもshapeは変わらず、例では両方とも`(2,)`です。

関連：[補足Cの一般の2x2行列積](#numpy-matrix)

```python
matrix_values = np.arange(4).reshape(2, 2)
vector_values = np.array([1, 2])
print(matrix_values)
print(matrix_values @ vector_values)
print(matrix_values.T)
print(vector_values.shape, vector_values.T.shape)
```

小数を保持したい配列は、作成時に`dtype=float`を指定します。ビット幅や大きな値の範囲についてはここでは扱わず、小さい値で代入時の違いだけを確認します。

```python
standard_integer_values = ar.array("i", [1, 2])
try:
    standard_integer_values[0] = -2.5
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

### 10. 相互変換とコピー（配布PDF pp.31–32）

例の`np.array(source_standard)`は標準`array`からNumPy配列を作り、ここでは独立した配列になるので、変換先の書き換えは元に反映されません。NumPy配列の`.tolist()`から`array.array`を作る例も示します。

2次元配列の`.tolist()`は入れ子のリストです。標準`array`へ移すときは、例のように平坦化して1次元の値にしますが、その時点で元のshape情報は失われます。形を再構成する必要がある場合はshapeを別途保ちます。NumPyの関数には`array_like`を受け取るものがあり、`np.sin`には標準`array`を渡せます。返ってくる値は`ndarray`です。変換方法によって共有やコピーの条件は異なるため、すべてのNumPy変換がコピーだとは一般化しません。

`flatten()`は、要素を1次元に並べた独立したコピーを返します。

関連：[7. 切り出した配列はコピーか共有か](#appendix-a7)

```python
source_standard = ar.array("d", [1, 2, 3])
converted_numpy = np.array(source_standard)
converted_numpy[0] = 99
print(source_standard.tolist())
restored_standard = ar.array("d", converted_numpy.tolist())
print(restored_standard.tolist())
```

```python
matrix_numpy = np.array([[1.0, 2.0], [3.0, 4.0]])
flat_standard = ar.array("d", matrix_numpy.flatten().tolist())
print(flat_standard.tolist())
```

### 11. 物理の計算で使い分ける（配布PDF pp.33–34）

静止した物体が原点から下向きを正として自由落下し、空気抵抗を無視する近似を考えます。重力加速度は`g = 10.0 m/s^2`、時刻は`t >= 0`です。位置は`x(t) = 1/2 g t^2 = 5.0 t^2`なので、`t = 0, 1, 2 s`で`x = 0, 5, 20 m`となります。

標準`array`の例では`for`で時刻を順に読み、計算した位置を`append`します。NumPyでは同じ式を配列全体に適用します。1次元の同型値を保管したい場合や標準ライブラリだけで処理したい場合は`array.array`が便利です。同じ数式を多数の値に適用する処理、多次元配列、行列計算には`ndarray`が便利です。この例は使い分けを示すもので、実行速度は比較していません。

関連：[本編の自由落下](#main-body-4) · [本編の回転](#main-body-13)

```python
times_standard = ar.array("d", [0.0, 1.0, 2.0])
positions_standard = ar.array("d")
for time_value in times_standard:
    positions_standard.append(5.0 * time_value ** 2)
times_numpy = np.array(times_standard)
positions_numpy = 5.0 * times_numpy ** 2
print(positions_standard.tolist())
print(positions_numpy)
```

## 補足B：Pythonのlist・tuple・set・dict

組込みコンテナの特徴を短い例で確認します。listとtupleは順序を保ち、setは重複を除き、dictはキーと値を対応づけます。例では本編の変数を使いません。

### list：順序があり、要素を変更できる

リストは数値や文字列など異なる型を混在できます。添字は0から始まり、`+`で結合、`len`で長さ、`in`で要素の有無を調べます。

```python
list_numbers = [1, 2, 3]
list_words = ["Hello", " ", "world"]
print(list_numbers[0], list_words[2], len(list_numbers), 2 in list_numbers)
list_numbers[0] = "A"
list_numbers.append("X")
for item in list_numbers:
    print(item)
print(list_numbers + list_words)
```

### tuple：順序があり、要素を変更できない

添字への代入はTypeErrorです。例外を捕捉して型名とメッセージを表示するため、Run Allを続けられます。タプル全体を新しい値で置き換えたり、複数の変数をまとめて交換したりできます。

```python
tuple_values = (1, 2, 3)
print(tuple_values)
try:
    tuple_values[0] = "A"
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
tuple_values = ("A", tuple_values[1], tuple_values[2])
left_value, right_value = 1, 2
left_value, right_value = right_value, left_value
print(tuple_values, left_value, right_value)
```

### set：重複を除き、集合演算ができる

集合は要素の順序を保証しません。次の例では表示のために`sorted`で並べています。`|`は和集合、`&`は共通部分、`^`は片方だけに含まれる要素（対称差）です。

```python
set_left = {1, 2, 3, 3, 4}
set_right = set([3, 1, 5, 5, 5, 4])
print(sorted(set_left), sorted(set_right))
print("和集合:", sorted(set_left | set_right))
print("共通部分:", sorted(set_left & set_right))
print("対称差:", sorted(set_left ^ set_right))
```

### dict：キーと値を対応づける

辞書はキーを追加した順序を保持します。キーには整数や文字列などhashableな値を使えますが、リストは使えません。次の例では物理定数の名前をキーにし、値の更新・追加と取り出しを試します。

```python
# hbarはJ s、cはm/s、eはC。c=1.0は更新例を示す仮の値
constants = {"hbar": 1.055e-34, "c": 1.0}
print(constants["hbar"])
constants["c"] = 2.998e+8
constants["e"] = 1.602e-19
for key, value in constants.items():
    print(key, value)
```

### ソート

`np.sort`は並べ替えた新しい配列を返します。配列の`.sort()`はその配列を直接変更します。ここでは比較のため、`.copy()`で作った配列を変更します。

```python
sort_input = np.array([2, 1, 5, 3, 7, 4, 6, 8])
sort_result = np.sort(sort_input)
in_place_sorted = sort_input.copy()
in_place_sorted.sort()
print(sort_result)
print(in_place_sorted)
print(sort_input)  # 元の配列は変わらない
```

### 条件、where、nonzero

条件式は真偽値の配列を作ります。np.nonzeroとnp.where(condition)は条件を満たす添字を返し、np.where(condition, x, y)は条件ごとに値を選んだ新しい配列を返します。

条件配列や整数配列を添字に使うと、必要な要素をまとめて選べます。添字への直接代入は元の配列を変更するので、最後の例では`.copy()`を作ってから負の値を0へ置き換えます。

```python
condition_values = np.array([-2, -1, 0, 1, 2, 3])
positive_condition = condition_values > 0
print(np.nonzero(positive_condition))
print(np.where(positive_condition))
print(np.where(positive_condition, condition_values, 0))
print(condition_values[positive_condition])
print(condition_values[np.array([0, 2, 4])])
masked_values = condition_values.copy()
masked_values[masked_values < 0] = 0
print(masked_values)
```

### 内積、外積（outer product）、ブロードキャスト

1次元配列の`np.dot`は対応する成分の積を足した内積です。`vector_left[None, :]`はshape `(1, 3)`、`vector_right[:, None]`はshape `(3, 1)`の配列を作ります。NumPyは大きさ1の軸を広げるようにして計算でき、これをブロードキャストと呼びます。

左のベクトルを$a$、右を$b$とすると、下の積の成分は$o_{ij} = b_i a_j$です。結果は3×3の行列になります。ここでの外積はouter productで、3次元ベクトルのcross productとは異なります。

```python
vector_left = np.array([1, 2, 3])
vector_right = np.array([4, 5, 6])
row_values = vector_left[None, :]
column_values = vector_right[:, None]
print("内積:", np.dot(vector_left, vector_right))
print("ブロードキャストによる和:")
print(row_values + column_values)
print("外積:")
print(row_values * column_values)
```

### einsumによる縮約

以下の関係は、すべて和を明示して書きます。e_jの積では添字iが3回現れるため、通常のEinstein規約として省略せず、和を明示します。

$$
\begin{aligned}
c &= \sum_i a_{ii} \\
d_{ik} &= \sum_j a_{ij} b_{jk} \\
e_j &= \sum_i a_{ii} b_{ij} \\
f &= \sum_{i,j,k} a_{ij} b_{kj} a_{ki}
\end{aligned}
$$

コードの`einsum_a`、`einsum_b`が式の$a$、$b$に対応します。返される$c$、$d$、$e$、$f$を順に確認します。

```python
import numpy as np

einsum_rng = np.random.default_rng(42)
einsum_a = einsum_rng.standard_normal((5, 5))
einsum_b = einsum_rng.standard_normal((5, 5))
trace_value = np.einsum("ii", einsum_a)
matrix_product = np.einsum("ij,jk->ik", einsum_a, einsum_b)
diagonal_weighted_sum = np.einsum("ii,ij->j", einsum_a, einsum_b)
scalar_contraction = np.einsum("ij,kj,ki->", einsum_a, einsum_b, einsum_a)
trace_value, matrix_product, diagonal_weighted_sum, scalar_contraction
```
