## NumPyを使ってみよう

NumPyは、数値計算の道具を集めたライブラリです。

- `import numpy as np`：NumPyを読み込み、短い名前 `np` で使います。
- `np.sqrt(9)`：NumPyの平方根を求める関数 `sqrt` に、値 `9` を渡します。関数名に続く括弧の中に、入力する値を書きます。
- `print(...)`：計算結果を表示します。

Colabを再起動したら、読み込みのセルをもう一度実行します。円周率は `np.pi`、自然対数の底 $e$ は `np.e` で使えます。

## 三角関数を計算しよう

`np.sin(x)`・`np.cos(x)`・`np.tan(x)` で、正弦・余弦・正接を計算します。

**角度の単位はラジアンです。** $\pi$ ラジアンが $180^\circ$、$\pi/2$ ラジアンが $90^\circ$ に対応します。

```python
print(np.sin(np.pi / 2))
print(np.cos(0))
print(np.tan(np.pi / 4))
```

### 度で指定したいとき

`np.deg2rad(30)` で $30^\circ$ をラジアンに変換してから、三角関数に渡します。$\sin 30^\circ$ は約0.5になります。

`np.sin(30)` は **30ラジアン** の計算です。比べてみましょう。

```python
angle = np.deg2rad(30)
print(np.sin(angle))
print(np.sin(30))
```

### 小さなずれにも注目

理論上 $\sin \pi=0$ ですが、円周率の近似や計算の丸めにより、小さなずれが出ることがあります。`e-16` は「$10^{-16}$ 倍」という数の表記です。

```python
print(np.sin(np.pi))
```

## 指数関数と対数関数

$10^3=1000$ なので、$\log_{10}1000=3$ です。

```python
print(np.log10(1000))
```

| 数学での表記 | NumPyの書き方 |
|---|---|
| $e^x$ | `np.exp(x)` |
| $\ln x=\log_e x$ | `np.log(x)` |
| $\log_{10}x$ | `np.log10(x)` |
| $\log_2 x$ | `np.log2(x)` |

**`np.log` は、底が $e$ の自然対数です。** 実数の範囲では、対数の入力は正の値にします。

```python
print(np.exp(0))
print(np.log(np.e))
print(np.log10(100))
print(np.log2(8))
```

## 配列でまとめて計算しよう

配列は、複数の値をまとめて扱うためのものです。`np.array([1, 4, 9])` は、1・4・9を要素とする配列を作ります。`np.sqrt(x)` は、各要素の平方根を計算します。

```python
x = np.array([1, 4, 9])
print(np.sqrt(x))
```

`np.linspace(開始, 終了, 個数)` で、両端を含む等間隔の値を作ります。下の例では、0から1までの5個の値です。

配列の詳しい使い方は11/10の授業で扱います。

```python
x = np.linspace(0, 1, 5)
print(x)
```

## 数式をコードにしてみよう

$y=e^{-x}\sin x$ を、$x=\pi/2$ で計算します。数式で省略される掛け算も、コードでは `*` を書きます。

```python
x = np.pi / 2
y = np.exp(-x) * np.sin(x)
print(y)
```

`x` の値を変えたら、`y` の計算も実行し直します。`x` が配列なら、同じ式で各要素をまとめて計算できます。次の例は、0から $2\pi$ までの9個の点で計算します。

```python
x = np.linspace(0, 2 * np.pi, 9)
y = np.exp(-x) * np.sin(x)
print(y)
```

$\sin^2 x+\cos^2 x=1$ を、好きな `x` で確かめましょう。`**2` は2乗です。
