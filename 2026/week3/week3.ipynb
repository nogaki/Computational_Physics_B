## 関数：重複する処理をまとめる
本文4ページ

同じ計算が繰り返されるときは、関数にまとめられます。ここでは1〜10と1〜30の総和をそれぞれ求め、足します。

```python
sum1 = 0
for i in range(10):
    sum1 = sum1 + (i + 1)

sum2 = 0
for j in range(30):
    sum2 = sum2 + (j + 1)

print(sum1 + sum2)
```

## 関数で1からnの総和を求める
本文5ページ

引数`n`を受け取り、1から`n`までを足した値を返す関数にします。

```python
def sum_one2n(n):
    total = 0
    for i in range(n):
        total = total + (i + 1)
    return total

sum1 = sum_one2n(10)
sum2 = sum_one2n(30)

print(sum1 + sum2)
```

## printとreturnの違い
本文7ページ

```python
def sum_all(x, y, z):
    tmp = x + y
    return (tmp + z)

l = 3
m = 2
n = 5

print(sum_all(l, m, n))
```

## 引数と物理パラメータ
本文9ページ

運動エネルギーは$E=mv^2/2$です。`m`は質量[kg]、`v`は速度[m/s]、返り値`E`の単位は[J]です。関数呼び出しの`2.0`が`m`、`3.0`が`v`に入ります。

```python
def kinetic_energy(m, v):
    return 0.5 * m * v ** 2

print(kinetic_energy(2.0, 3.0))
```

## 複数の返り値：自由落下
本文10ページ

空気抵抗を無視し、静止状態から落下するとします。開始位置を$x=0$、下向きを正とします。教材の計算例では重力加速度を$g=10.0\ \mathrm{m/s^2}$と近似します。落下開始からの経過時間$t\geq0$ [s]、位置$x$ [m]、速度$v$ [m/s]として、$x=gt^2/2$、$v=gt$です。

```python
def free_fall(t):
    g = 10.0
    x = 0.5 * g * t ** 2
    v = g * t
    return x, v
```

## タプルを受け取る
本文11ページ

カンマでまとめた値の組をタプルといいます。`return x, v`は2つの値を順にまとめて返します。受け取る変数の数は返した値の数に合わせます。$t=3.0$ sでは$x=45.0$ m、$v=30.0$ m/sです。

```python
state = free_fall(3.0)
print(state)
x, v = state
print(x, v)
```

## 関数を組み合わせる
本文12ページ

`free_fall`の返り値の速度を`kinetic_energy`の引数に渡します。質量2.0 kg、落下開始から3.0 s後の運動エネルギーを求めます。この例では出力が900.0で、単位は[J]です。

```python
m = 2.0
x, v = free_fall(3.0)
energy = kinetic_energy(m, v)
print(energy)
```

## 漸化式を関数で表す
本文14ページ

$a_1=1$、$a_{n+1}=2a_n+3$を関数で繰り返し計算し、$a_{10}$を求めます。一般項は$a_n=4\cdot2^{n-1}-3$です。

```python
def next_term(a):
    return 2 * a + 3

a = 1
for _ in range(9):
    a = next_term(a)

print(a)
```

## 最大公約数：while文を使う
本文15ページ

ユークリッドの互除法では、余りが0になるまで割る数と余りを入れ替えます。次の例は正の整数を対象にしています。

```python
def gcd(n, m):
    while m != 0:
        r = n % m
        n = m
        m = r

    return n
```

```python
gcd(120, 12)
```

## 発展：再帰関数で最大公約数を求める
本文17ページ

再帰関数は、自分自身を呼び出す関数です。以下はユークリッドの互除法を再帰処理で書いた例です。

```python
def gcd_recursive(n, m):
    if m == 0:
        return n
    return gcd_recursive(m, n % m)
```

```python
gcd_recursive(120, 12)
```
